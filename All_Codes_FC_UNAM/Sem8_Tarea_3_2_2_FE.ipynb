```python
import numpy as np
import matplotlib.pyplot as plt

# Constantes físicas arbitrarias para ilustrar la forma funcional
A = 1  # área arbitraria (se cancela o escala linealmente)
rho = 1  # densidad arbitraria
sigma = 1  # tensión superficial arbitraria
k_B = 1.38e-23  # constante de Boltzmann (en J/K)
hbar = 1.054e-34  # constante de Planck reducida (en J·s)

# Definimos la constante prefactor general (sin unidades)
C_prefactor = A * (rho / sigma)**(4/3) * k_B**(7/3) / (2 * np.pi) * (2/3)

# Usamos funciones especiales para valores reales
from scipy.special import gamma, zeta

# Valor numérico de la integral
gamma_val = gamma(7/3)
zeta_val = zeta(7/3)

# Constante completa
C = C_prefactor * gamma_val * zeta_val

# Rango de temperaturas (en unidades arbitrarias)
T = np.linspace(1e-2, 10, 500)

# Calor específico a bajas temperaturas (forma teórica)
C_V_lowT = (7/3) * C * T**(4/3)

# Calor específico a altas temperaturas (línea constante)
C_V_highT = np.ones_like(T) * k_B * 100  # asumimos 100 modos accesibles como ejemplo

# Gráfica combinada
plt.figure(figsize=(10, 6))
plt.plot(T, C_V_lowT, label=r'$C_V \propto T^{4/3}$ (bajas T)', linewidth=2)
plt.plot(T, C_V_highT, '--', label=r'$C_V \approx \text{cte}$ (altas T)', linewidth=2, color='orange')
plt.axvline(x=2, color='gray', linestyle=':', label='Transición cualitativa')
plt.xlabel('Temperatura (T)')
plt.ylabel('Calor específico $C_V$')
plt.title('Comportamiento del calor específico $C_V$ de ondas superficiales en un líquido')
plt.legend()
plt.grid(True)
plt.yscale('log')
plt.xscale('log')
plt.tight_layout()
plt.show()
```
