```python
import numpy as np
import matplotlib.pyplot as plt

# Constantes
P0 = 101325  # Presión a nivel del mar en Pa
L = 0.0065   # Gradiente térmico de temperatura en K/m
T0 = 288.15  # Temperatura a nivel del mar en K
g = 9.81     # Aceleración debido a la gravedad en m/s²
M = 0.029    # Masa molar del aire en kg/mol
R = 8.314    # Constante de los gases ideales en J/(mol·K)

# Función para calcular la presión a una altitud h
def presion_atmosferica(h):
    return P0 * (1 - L * h / T0) ** (g * M / (R * L))

# Altitudes de medición (de 0 a 30 metros)
altitudes = np.linspace(0, 30, 30)

# Calcular presiones para las altitudes dadas
presiones = presion_atmosferica(altitudes)

# Graficar la relación de presión con la altitud
plt.figure(figsize=(8, 6))
plt.plot(altitudes, presiones, label="Presión atmosférica", color="b")
plt.xlabel("Altitud (m)")
plt.ylabel("Presión atmosférica (Pa)")
plt.title("Cambio de presión atmosférica con la altitud (hasta 30 metros)")
plt.grid(True)
plt.legend()
plt.show()

# Imprimir las presiones para cada altitud
for h, p in zip(altitudes, presiones):
    print(f"Altitud: {h:.1f} m, Presión: {p:.2f} Pa")
```
